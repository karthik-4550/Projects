## Clean Data

### Subtask:
Address missing values, incorrect data types, and any obvious inconsistencies or errors found during the initial inspection.


**Reasoning**:
I will clean and convert the data types of `discounted_price`, `actual_price`, `discount_percentage`, `rating`, and `rating_count` columns, and then impute missing values in `rating_count` as per the instructions. Finally, I will verify the changes with `df.info()` and `df.isnull().sum()`.



**Reasoning**:
The previous code failed because the 'rating' column contained an unhandled non-numeric string ('|') that could not be converted to float. I need to modify the cleaning step for the 'rating' column to convert all non-numeric values to NaN using `pd.to_numeric` with `errors='coerce'`, then fill those NaN values with the median rating before converting to float.



**Reasoning**:
The `AttributeError` indicates that the `discounted_price` column might not be entirely of string type, preventing the `.str` accessor from being used. To fix this, I need to explicitly convert the `discounted_price` and `actual_price` columns to string type using `.astype(str)` before applying string manipulations, and then proceed with the rest of the cleaning steps, ensuring robustness against prior partial conversions or mixed types.



```python
print('Cleaning discounted_price and actual_price columns...')
df['discounted_price'] = df['discounted_price'].astype(str).str.replace('₹', '', regex=False).str.replace(',', '', regex=False).astype(float)
df['actual_price'] = df['actual_price'].astype(str).str.replace('₹', '', regex=False).str.replace(',', '', regex=False).astype(float)
print('discounted_price and actual_price cleaned and converted to float.')

print('\nCleaning discount_percentage column...')
df['discount_percentage'] = df['discount_percentage'].astype(str).str.replace('%', '', regex=False).astype(int)
print('discount_percentage cleaned and converted to int.')

print('\nCleaning rating column...')
df['rating'] = df['rating'].replace('4.0 out of 5 stars', '4.0')
df['rating'] = pd.to_numeric(df['rating'], errors='coerce') # Convert to numeric, coercing errors to NaN
median_rating = df['rating'].median()
df['rating'].fillna(median_rating, inplace=True)
df['rating'] = df['rating'].astype(float)
print(f'rating cleaned, missing values imputed with median ({median_rating}), and converted to float.')

print('\nCleaning and imputing rating_count column...')
df['rating_count'] = df['rating_count'].astype(str).str.replace(',', '', regex=False)
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce') # Convert to numeric, coercing errors to NaN
median_rating_count = df['rating_count'].median()
df['rating_count'].fillna(median_rating_count, inplace=True)
df['rating_count'] = df['rating_count'].astype(int)
print(f'rating_count cleaned, missing values imputed with median ({median_rating_count}), and converted to int.')

print('\nVerifying data types and missing values after cleaning:')
df.info()
print('\nMissing values per column after cleaning:')
print(df.isnull().sum())
```

**Reasoning**:
The previous output included `FutureWarning` messages about using `inplace=True` with `fillna()`. To prevent these warnings and ensure correct behavior in future pandas versions, I will modify the code to assign the result of `fillna()` back to the DataFrame column instead of using the `inplace` parameter.



```python
print('Cleaning discounted_price and actual_price columns...')
df['discounted_price'] = df['discounted_price'].astype(str).str.replace('₹', '', regex=False).str.replace(',', '', regex=False).astype(float)
df['actual_price'] = df['actual_price'].astype(str).str.replace('₹', '', regex=False).str.replace(',', '', regex=False).astype(float)
print('discounted_price and actual_price cleaned and converted to float.')

print('\nCleaning discount_percentage column...')
df['discount_percentage'] = df['discount_percentage'].astype(str).str.replace('%', '', regex=False).astype(int)
print('discount_percentage cleaned and converted to int.')

print('\nCleaning rating column...')
df['rating'] = df['rating'].replace('4.0 out of 5 stars', '4.0')
df['rating'] = pd.to_numeric(df['rating'], errors='coerce') # Convert to numeric, coercing errors to NaN
median_rating = df['rating'].median()
df['rating'] = df['rating'].fillna(median_rating)
df['rating'] = df['rating'].astype(float)
print(f'rating cleaned, missing values imputed with median ({median_rating}), and converted to float.')

print('\nCleaning and imputing rating_count column...')
df['rating_count'] = df['rating_count'].astype(str).str.replace(',', '', regex=False)
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce') # Convert to numeric, coercing errors to NaN
median_rating_count = df['rating_count'].median()
df['rating_count'] = df['rating_count'].fillna(median_rating_count)
df['rating_count'] = df['rating_count'].astype(int)
print(f'rating_count cleaned, missing values imputed with median ({median_rating_count}), and converted to int.')

print('\nVerifying data types and missing values after cleaning:')
df.info()
print('\nMissing values per column after cleaning:')
print(df.isnull().sum())
```

## Perform Exploratory Data Analysis (EDA)

### Subtask:
Conduct exploratory data analysis to uncover patterns, relationships, and key statistics within the cleaned dataset. This includes descriptive statistics, distribution analysis, and identifying potential trends.


**Reasoning**:
To begin the exploratory data analysis, I will display descriptive statistics for all numerical columns to understand their central tendency, dispersion, and shape.



```python
print('Descriptive Statistics for Numerical Columns:')
print(df.describe())
```

**Reasoning**:
Following the instructions, I will now calculate the actual discount amount for each product and store it in a new column named `discount_amount` by subtracting `discounted_price` from `actual_price`.



```python
df['discount_amount'] = df['actual_price'] - df['discounted_price']
print('New column discount_amount created.')
print(df[['actual_price', 'discounted_price', 'discount_amount']].head())
```

**Reasoning**:
To identify the main product categories, I will extract the first part of the 'category' string before the first '|' and store it in a new column named `main_category`. Then, I will find and display the top 5 most frequent main categories.



```python
print('Extracting main_category from category column...')
df['main_category'] = df['category'].apply(lambda x: x.split('|')[0] if pd.notnull(x) else x)
print('main_category column created.')

print('\nTop 5 most frequent main categories:')
print(df['main_category'].value_counts().head(5))
```

**Reasoning**:
As per the instructions, I will now group the DataFrame by `main_category` and calculate the average `discounted_price`, `actual_price`, `discount_percentage`, `rating`, and `rating_count` for each main category.



```python
print('Average metrics by main category:')
category_avg = df.groupby('main_category')[['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']].mean().sort_values(by='actual_price', ascending=False)
print(category_avg)
```

**Reasoning**:
As per the instructions, I will now identify the top 10 products with the highest `rating_count` (most reviewed products) and display their `product_name`, `rating`, and `rating_count`.



```python
print('Top 10 products with the highest rating count:')
top_rated_products = df.sort_values(by='rating_count', ascending=False).head(10)
print(top_rated_products[['product_name', 'rating', 'rating_count']])
```

## Visualize Key Insights

### Subtask:
Generate plots to visually represent important findings or trends identified during the EDA, such as the distribution of discount percentages and average prices across main product categories.


**Reasoning**:
To visualize the distribution of discount percentages and average prices across main product categories, I will first import the necessary plotting libraries, `matplotlib.pyplot` and `seaborn`. Then, I will generate the required plots as instructed and ensure they are properly labeled and titled.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set_style('whitegrid')

# Plot 1: Distribution of Discount Percentages
plt.figure(figsize=(10, 6))
sns.histplot(df['discount_percentage'], kde=True, bins=20)
plt.title('Distribution of Discount Percentages')
plt.xlabel('Discount Percentage (%)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()

# Plot 2: Average Actual vs. Discounted Price by Main Category
plt.figure(figsize=(14, 7))
# Sort category_avg by actual_price in descending order
category_avg_sorted = category_avg.sort_values(by='actual_price', ascending=False)

# Create bar plot
category_avg_sorted[['actual_price', 'discounted_price']].plot(kind='bar', figsize=(14, 7))
plt.title('Average Actual vs. Discounted Price by Main Category')
plt.xlabel('Main Category')
plt.ylabel('Average Price (₹)')
plt.xticks(rotation=45, ha='right')
plt.legend(['Actual Price', 'Discounted Price'])
plt.tight_layout()
plt.show()
```
